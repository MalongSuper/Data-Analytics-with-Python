## Data Analytics

### Chapter 3: Data Manipulation (Filtering & Sorting)

Data manipulation is a fundamental step in data analysis, allowing you to transform and modify your dataset for more targeted insights. This chapter explores various techniques for manipulating data effectively.

- **Filtering**: Selecting a subset of your dataset based on certain criteria.

- **Sorting**: Arranging the rows of your dataset in a specific order based on one or more columns.

#### Generate Synthetic Data

```python
import numpy as np
import pandas as pd

# Reproducible randomness
np.random.seed(42)
```

```python
names = [
    "James", "Chris", "Adam", "Marian", "Peter",
    "Kevin", "Linda", "Emma", "Sophia", "Daniel",
    "Noah", "Olivia", "Lucas", "Grace", "Henry",
    "Liam", "Ava", "Mia", "Nathan", "Chloe"
]

majors = ["AI", "IT", "CSC", "SW"]

n = 20

df = pd.DataFrame({
    "Name": np.array(names),
    "Age": np.random.randint(18, 21, n),
    "Major": np.random.choice(majors, n),
    "GPA": np.round(np.random.uniform(2.0, 4.0, n), 2)
})

df
```

#### 3.1. Boolean Series

Boolean Series is a series of True/False values that can be used to filter data in a DataFrame. It is often created by applying a condition to a column or multiple columns in the DataFrame.

```python
# Return a Boolean series
df["GPA"] >= 3.5
```

When combining the boolean series with the DataFrame, it returns a new DataFrame that only includes the rows where the condition is True. 

```python
# returns only the rows whose corresponding Boolean value is True
df[df["GPA"] >= 3.5]
```

#### 3.2. Filtering Based on Comparison Operators

Some common comparison operators used for filtering include:

- `==` (equal to)

- `!=` (not equal to)

- `>` (greater than)

- `<` (less than)

- `>=` (greater than or equal to)

- `<=` (less than or equal to)

```python
# Example 1: Students with GPA greater than 3.5
df[df["GPA"] > 3.5]
```

```python
# Example 2: Students aged 20
df[df["Age"] == 20]
```

```python
# Example 3: Students who are not in the AI major
df[df["Major"] != "AI"]
```

#### 3.3. Filtering Based on Logical Conditions

Logical conditions can be combined using the following operators:

- `&` (and)

- `|` (or)

- `~` (not)

```python
# Return students who are in the AI major and have a GPA greater than 3.5.
df[(df["Major"] == "AI") & (df["GPA"] > 3.5)]
```

```python
# Return students who are either AI majors or have a GPA above 3.8.
df[(df["Major"] == "AI") | (df["GPA"] > 3.8)]
```

```python
# Return students who are not AI majors.
df[~(df["Major"] == "AI")]
```

#### 3.4. Filtering Based on Subset Membership

Three common methods for filtering based on subset membership include:

- Using the `isin()` method to filter rows based on whether a column's values are in a specified list.

- `any()` and `all()` methods to filter rows based on whether any or all of the specified conditions are met.

```python
# Return students majoring in AI or CSC.
df[df["Major"].isin(["AI", "CSC"])]
```

```python
# Return True if at least one student has a GPA greater than 3.9.
(df["GPA"] > 3.9).any()
```

```python
# If every generated student is at least 18 years old, the result is True
(df["Age"] >= 18).all()
```

#### 3.5. Filtering Based on String Conditions

String conditions can be used to filter rows based on text data. Some common string methods for filtering include:

- `str.contains()` to filter rows based on whether a column's values contain a specified substring.

- `str.startswith()` to filter rows based on whether a column's values start with a specified substring.

- `str.endswith()` to filter rows based on whether a column's values end with a specified substring.

```python
# Names containing "an"
df[df["Name"].str.contains("an")]
```

```python
# Example 2: Names starting with "A"
df[df["Name"].str.startswith("A")]
```

```python
# Example 3: Names ending with "a"
df[df["Name"].str.endswith("a")]
```

#### 3.6. Sorting by Column Values

When sorting a DataFrame by column values, you can specify one or more columns to sort by. The `sort_values()` method allows you to sort the DataFrame in ascending or descending order.

```python
# Example 1: Sort by GPA
df.sort_values(by="GPA")
```

```python
# Example 2: Sort in Descending Order
df.sort_values(by="GPA", ascending=False)
```

```python
# Example 3: Sort by Multiple Columns
df.sort_values(by=["Major", "GPA"])
```

```python
# Each column can also have its own sorting direction.
df.sort_values(by=["Major", "GPA"], ascending=[True, False])
```

#### 3.7. Sorting by Index

Sorting by index requires the data to have meaningful index labels. The `sort_index()` method allows you to sort the DataFrame by its index in ascending or descending order.

```python
df_index = df.set_index("Name")
df_index.sort_index()
```

#### 3.8. Duplicate Values in Sorting

Sorting can be improved by handling duplicate values. The `by` parameter in the `sort_values()` method allows you to specify multiple columns to sort by. 

When same values occur, immediately use the next column to sort by. This ensures that the sorting is stable and consistent.

```python
df.sort_values(by=["Major", "Age", "GPA", "Name"])
```
