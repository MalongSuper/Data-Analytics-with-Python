### Chapter 4: Data Manipulation (Replacing & Mapping)

Other than filtering and sorting, data manipulation also includes techniques for replacing existing values with new ones or creating a new dataset based on existing data. This chapter explores various techniques for replacing and mapping data effectively.

**Replacing and Mapping are techniques used in data manipulation to modify or transform data values in a dataset. These techniques are essential for cleaning, standardizing, and preparing data for analysis.**

While they are similar, they serve different purposes:

- **Replacing**: This involves substituting existing values in a dataset with new values. It is often used to correct errors, standardize data, or handle missing values. For example, replacing all instances of "N/A" with `NaN` or changing categorical values like "Yes" and "No" to `1` and `0`, respectively.

- **Mapping**: This involves creating a new dataset or transforming existing data based on a defined mapping or relationship. Mapping is often used to convert categorical variables into numerical representations, create new features, or apply functions to transform data. For example, mapping a list of country names to their corresponding ISO codes or applying a function to calculate the square of numerical values.

#### Generate Synthetic Data

```python
import numpy as np
import pandas as pd

# Reproducible randomness
np.random.seed(42)
```

```python
names = [
    "James", "Chris", "Adam", "Marian", "Peter",
    "Kevin", "Linda", "Emma", "Sophia", "Daniel",
    "Noah", "Olivia", "Lucas", "Grace", "Henry",
    "Liam", "Ava", "Mia", "Nathan", "Chloe"
]

majors = ["AI", "IT", "CSC", "SW"]

n = 20

df = pd.DataFrame({
    "Name": np.array(names),
    "Age": np.random.randint(18, 21, n),
    "Major": np.random.choice(majors, n),
    "GPA": np.round(np.random.uniform(2.0, 4.0, n), 2)
})

df
```

```python
# Create two temporary datasets for this section
df_replace = df.copy()
df_map = df.copy()
```

#### 4.1. Replace a Single Value

You can replace a single value in a DataFrame using the `replace()` method. This method allows us to specify the value we want to replace and the new value we want to use.

```python
df_replace["Major"] = df_replace["Major"].replace("CSC", "Computer Science")

df_replace
```

#### 4.2. Replace Multiple Values

You can replace multiple values in a DataFrame by passing a dictionary to the `replace()` method. The keys of the dictionary represent the values to be replaced, and the corresponding values represent the new values.

```python
# A dictionary can replace several values at once.
df_replace["Major"] = df_replace["Major"].replace({
    "AI": "Artificial Intelligence",
    "IT": "Information Technology",
    "CSC": "Computer Science",
    "SW": "Software Engineering"
})

df_replace
```

```python
# You can even replace values across the entire DataFrame if needed.
df_replace1 = df.copy()
df_replace1 = df_replace1.replace({"CSC": "CS", "SW": "SE"})
df_replace1
```

#### 4.3. Basic Mapping

Mapping is similar to replacing, but it is often used to create a new dataset or transform existing data based on a defined mapping or relationship. 

The `map()` method allows you to apply a function or a mapping dictionary to a Series, transforming its values accordingly.

```python
major_map = {
    "AI": "Artificial Intelligence",
    "IT": "Information Technology",
    "CSC": "Computer Science",
    "SW": "Software Engineering"
}

df_map["Major"] = df_map["Major"].map(major_map)

df_map
```

#### 4.4. Label Encoding Example

Mapping can also be used for label encoding, where categorical values are converted into numerical labels. This is particularly useful for machine learning algorithms that require numerical input.

```python
df_map = df.copy()

major_code = {
    "AI": 0,
    "IT": 1,
    "CSC": 2,
    "SW": 3
}

df_map["Major_Code"] = df_map["Major"].map(major_code)
df_map = df_map.drop(columns=["Major"])

df_map
```

#### 4.5. Mapping with `unique()`

```python
df_map = df.copy()
unique_major = df["Major"].unique()

mapping = {
    value: index
    for index, value in enumerate(unique_major)
}

df_map["Major_Code"] = df_map["Major"].map(mapping)
df_map = df_map.drop(columns=["Major"])

df_map
```

#### 4.6. Replacing Missing Values

Missing Values can be anything:

- Some common values like `NaN`; `None` ; `"N/A"` ; `"Unknown"` ;  

- An empty string `""` ; 

- A blank space `" "` ;  

- A placeholder value such as 0 (when 0 is not a valid value)

```python
import numpy as np
import pandas as pd

df = pd.DataFrame({
    "Name": [
        "James", "Chris", "Adam", "Marian", "Peter",
        "Kevin", "Linda", "Emma", "Sophia", "Daniel"
    ],
    "Major": [
        "AI", "IT", "Unknown", "CSC", None,
        "", "SW", "N/A", "AI", "CSC"
    ],
    "Age": [
        19, 20, 0, 18, np.nan,
        19, None, 20, 18, 0
    ],
    "GPA": [
        3.60, np.nan, 2.90, 0,
        3.10, None, "N/A", 3.80,
        "", 3.20
    ]
})

df.set_index("Name", inplace=True)

df
```

#### 4.7. Standardizing Missing Values

You can use the `replace()` method to standardize missing values in a DataFrame. This is particularly useful when dealing with datasets that have inconsistent representations of missing data. 

By replacing various representations of missing values with a single standard value (e.g., `NaN`), you can simplify data cleaning and analysis.

```python
df_imputed = df.replace({
    "Unknown": np.nan,
    "N/A": np.nan,
    "": np.nan,
    " ": np.nan,
    0: np.nan
})

df_imputed
```

#### 4.8. Imputing Missing Values after Standardization

Now we can simply impute the missing values using the `fillna()` method. This method allows us to fill in missing values with a specified value or a calculated statistic (e.g., mean, median, mode) of the column.

```python
# $eplace missing GPAs with the column mean
df_imputed["GPA"] = df_imputed["GPA"].fillna(df_imputed["GPA"].mean())

# Replace missing majors with "Undeclared"
df_imputed["Major"] = df_imputed["Major"].fillna("Undeclared")

# Replace missing ages with the median age
df_imputed["Age"] = df_imputed["Age"].fillna(df["Age"].median())

df_imputed
```
