## Data Analytics

### Chapter 5: Data Modification 

Modifying data is a fundamental aspect of data analytics, allowing analysts to clean, transform, and prepare datasets for analysis. This chapter covers various techniques for modifying data effectively.

- **Rename**: Rename columns or indexes in a dataset to make them more descriptive or consistent with naming conventions. This is often done to improve readability and understanding of the data.

- **Set or Reset Index**: Change the index of a dataset to a different column or reset it to the default integer index. This can be useful for reordering data or preparing it for analysis.

- **Insert, Delete, and Update Values**: Modify the contents of a dataset by adding new rows, removing existing ones, or updating specific values. This is crucial for maintaining accurate and relevant data.

#### Generate Synthetic Data

```python
import numpy as np
import pandas as pd

# Reproducible randomness
np.random.seed(42)
```

```python
names = [
    "James", "Chris", "Adam", "Marian", "Peter",
    "Kevin", "Linda", "Emma", "Sophia", "Daniel",
    "Noah", "Olivia", "Lucas", "Grace", "Henry",
    "Liam", "Ava", "Mia", "Nathan", "Chloe"
]

majors = ["AI", "IT", "CSC", "SW"]

n = 20

df = pd.DataFrame({
    "Name": np.array(names),
    "Age": np.random.randint(18, 21, n),
    "Major": np.random.choice(majors, n),
    "GPA": np.round(np.random.uniform(2.0, 4.0, n), 2)
})

df
```

#### 5.1. Retrieving Columns

You can retrieve specific columns from a dataset using indexing or the `loc` and `iloc` methods. This allows you to focus on the relevant data for your analysis.

```python
print(df.columns)  # returns an Index object containing all column names
print(df.columns.to_list()) # converts the Index object into a regular Python list
```

#### 5.2. Renaming Existing Columns

Renaming columns is a common task in data manipulation, especially when working with datasets that have unclear or inconsistent column names. You can use the `rename()` method to rename columns in a DataFrame.

```python
# The recommended way to rename columns is with df.rename()
df_rename = df.copy()
df_rename = df_rename.rename(columns={"GPA": "Grade"})

df_rename
```

```python
# Rename multiple columns simultaneously.
df_rename = df_rename.rename(columns={
    "Name": "Student",
    "Major": "Program",
    "GPA": "Grade"
})

df_rename
```

```python
# You can also replace all column names directly.
df_rename = df.copy()
df_rename.columns = ["Student", "Age", "Program", "Grade"]
df_rename 
```

#### 5.3. Setting an Existing Column as the Index

To set an index, you can use the `set_index()` method. This allows you to designate a specific column as the index of the DataFrame, which can be useful for data organization and retrieval.

```python
df_rename = df.copy()
df_rename = df_rename.set_index("Name")

df_rename
```

#### 5.4. Restoring the Default Index

When restoring the default index of a DataFrame, you can use the `reset_index()` method. This is useful when you want to revert to the original integer-based index after setting a different column as the index.

```python
df_rename = df_rename.reset_index()
df_rename
```

#### 5.5. Inserting New Columns

Inserting a new column into a DataFrame can be done by assigning a new value to a column name that does not yet exist. This allows you to add additional data or features to your dataset for analysis.

```python
df_new = df.copy()
df_new["Scholarship"] = False

df_new
```

```python
df_new["Year"] = [1, 2, 3, 4] * 5
df_new
```

#### 5.6. Inserting New Rows

Inserting new rows into a DataFrame can be accomplished using the `loc` method. This allows you to add new observations or records to your dataset, which can be useful for expanding your analysis or incorporating new data points.

Note that when using `loc`, make sure the list of values you provide matches the number of columns in the DataFrame to avoid errors.

```python
df_new.loc[len(df_new)] = ["Alice", 19, "AI", 3.85, False, 3]
df_new.tail(5)
```

#### 5.7. Deleting Columns

Delete columns from a DataFrame using the `drop()` method. This is useful for removing irrelevant or redundant data, helping to streamline your dataset for analysis.

```python
# To remove one or more columns, use drop().
df_drop = df_new.drop(columns=["Major"])
df_drop
```

```python
# To delete multiple columns
df_drop = df_new.drop(columns=["Age", "Major", "Scholarship"])
df_drop
```

```python
# Removing a Column with pop()
major = df_new.pop("Major")
major
```

#### 5.8. Deleting Rows

Deleting rows from a DataFrame can be done using the `drop()` method with the appropriate index labels. This is useful for removing specific observations that may be outliers or irrelevant to your analysis.

```python
# Rows are removed by specifying their index labels.
df_drop_index = df_new.drop(index=0)
df_drop_index
```

```python
# Delete multiple rows.
df_drop_index = df_new.drop(index=[0, 3, 5])
df_drop_index
```

#### 5.9. Updating an Entire Column

We can update the entire column by replacing it with a new list of values. This is useful when you need to modify all entries in a column based on new data or calculations.

```python
# Assigning a value to a column replaces every value in that column.
df_update = df_new.copy()
df_update["Major"] = "CSC"

df_update
```

```python
# You can also assign a list or Series.
df_update["Age"] = np.random.randint(18, 21, len(df_update))
df_update
```

#### 5.10. Updating Specific Rows

Updating specific rows in a DataFrame can be done by using boolean indexing or the `loc` method to target specific rows based on conditions. This allows for precise modifications to the dataset.

```python
# award a scholarship to students with a GPA of at least 3.5
df_update["Scholarship"] = False
df_update.loc[df_update["GPA"] >= 3.5, "Scholarship"] = True

df_update[df_update["GPA"] >= 3.5]
```

#### 5.11. Updating a DataFrame Using Filtering

Filtering is commonly used to update a DataFrame based on specific conditions. By applying filters, you can selectively modify rows that meet certain criteria, enabling targeted data manipulation.

Note that it only returns the rows that meet the specified conditions, allowing for efficient updates to the dataset without affecting unrelated data.

```python
# Filtering is commonly used to create a new DataFrame containing only selected rows.
high_gpa = df_new[df_new["GPA"] >= 3.5]
high_gpa
```
