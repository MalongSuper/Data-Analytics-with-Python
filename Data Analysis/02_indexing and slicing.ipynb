## Data Analytics

### Chapter 2: Data Subsetting (Indexing & Slicing)

Data subsetting is a fundamental step in data analysis, allowing you to focus on specific portions of your dataset for more targeted insights. This chapter explores various techniques for subsetting data effectively.

- **Length, Shape and Size**: Understanding the dimensions of your dataset to determine how many rows and columns it contains.
- **Indexing**: Methods to access specific rows and columns in your dataset using labels or integer-based positions.
- **Slicing**: Techniques to extract a range of rows or columns from your dataset for focused analysis.

#### Generate Synthetic Data

```python
import numpy as np
import pandas as pd

# Reproducible randomness
np.random.seed(42)
```

```python
names = [
    "James", "Chris", "Adam", "Marian", "Peter",
    "Kevin", "Linda", "Emma", "Sophia", "Daniel",
    "Noah", "Olivia", "Lucas", "Grace", "Henry",
    "Liam", "Ava", "Mia", "Nathan", "Chloe"
]

majors = ["AI", "IT", "CSC", "SW"]

n = 20

df = pd.DataFrame({
    "Name": np.array(names),
    "Age": np.random.randint(18, 21, n),
    "Major": np.random.choice(majors, n),
    "GPA": np.round(np.random.uniform(2.0, 4.0, n), 2)
})

df
```

#### 2.1. Data Length, Shape and Size

You can use the `len()`, `shape`, and `size` attributes to get information about the dimensions of your dataset.

```python
# Retrieve Length, Size, and Shape of data
print("Number of Samples:", len(df))
print("Number of Columns:", len(df.columns))
```

```python
# Counting Unique Values
print("Unique Values in Major column:", df["Major"].unique())
print("Number of Unique Values in Major column:", len(df["Major"].unique()))
```

```python
# Calculate average
# You can also use sum(df["GPA"]) / len(df) 
average_gpa = sum(df["GPA"]) / len(df["GPA"]) 
print('Average GPA:', average_gpa)
```

```python
print("DataFrame Shape:", df.shape)
print(" + Rows:", df.shape[0])
print(" + Columns:", df.shape[1])
print("DataFrame Size:", df.size)
```

#### 2.2. Data Indexing and Slicing

Some common types of indexing and slicing include:

- Using column names to access specific columns.

- Integer-based indexing to access specific rows or columns.

- Label-based indexing to access specific rows or columns using labels.

```python
# Size of Rows and Columns
print("Size of Column GPA:", df["GPA"].size)
print("Size of Rows:", df.iloc[0].size)
```

```python
# Retrieve the first five students
df.iloc[:5]
```

```python
# Retrieve rows 10–19
df.iloc[10:20]
```

```python
# Retrieve the Name and GPA columns
df[["Name", "GPA"]]
```

```python
# Retrieve the first five students and only the Name and Major columns
df.iloc[:5, [0, 2]]
```

```python
# Retrieve all GPAs
df.loc[:, "GPA"]
```

```python
# Retrieve all columns except Name
df.iloc[:, 1:]
```
