## Data Analytics

### Chapter 13: Exploratory Data Analysis - Part I

Exploratory Data Analysis (EDA) is the process of investigating and summarizing data using statistical methods and visualizations. 

The objective is to understand the underlying structure of the data, discover patterns, identify unusual observations, and explore relationships between variables before performing more advanced analysis.

#### Generate Synthetic Data

```python
import numpy as np
import pandas as pd

np.random.seed(42)
```

```python
n = 20

names = [
    "James", "Chris", "Adam", "Marian", "Peter",
    "Kevin", "Linda", "Emma", "Sophia", "Daniel",
    "Noah", "Olivia", "Lucas", "Grace", "Henry",
    "Liam", "Ava", "Mia", "Nathan", "Chloe"
]

df = pd.DataFrame({
    "Name": names,
    "Age": np.random.randint(18, 22, n),
    "Major": np.random.choice(["AI", "CSC", "IT", "SW"], n),
    "GPA": np.round(np.random.uniform(2.0, 4.0, n), 2)
})

df
```

#### Retrieve Basic Information

- **`df.info()`** displays the dataset structure, including the number of rows, columns, data types, and missing values.
- **`df.describe()`** summarizes numerical columns by reporting statistics such as count, mean, standard deviation, minimum, quartiles, and maximum.

```python
df.info()
```

```python
df.describe()
```

```python
df.describe(include='all')
```

```python
# create a second GPA column that will be used 
# to demonstrate arithmetic operations.
df_temp = df.copy()

df_temp["GPA2"] = np.round(np.random.uniform(2.0, 4.0, len(df)), 2)
df_temp[["Name", "GPA", "GPA2"]]
```

#### 13.1. Basic Statistical Operations

Pandas provide several functions allow us to perform arithmetic on columns, summarize numerical data, and identify the largest or smallest values.

- **Arithmetic Operations**: Pandas supports element-wise arithmetic between columns.

```python
# Addition
df_temp["GPA_Add"] = df_temp["GPA"].add(df_temp["GPA2"])

# Subtraction
df_temp["GPA_Sub"] = df_temp["GPA"].sub(df_temp["GPA2"])

# Multiplication
df_temp["GPA_Mul"] = df_temp["GPA"].multiply(df_temp["GPA2"])

# Division
df_temp["GPA_Div"] = df_temp["GPA"].divide(df_temp["GPA2"])

df_temp[["GPA", "GPA2", "GPA_Add", "GPA_Sub", "GPA_Mul", "GPA_Div"]]
```

- **Maximum and Minimum**: The maximum and minimum values can be obtained using `max()` and `min()`.

```python
print("Maximum GPA:", df["GPA"].max())
print("Minimum GPA:", df["GPA"].min())
```

- **Largest and Smallest Values**: To retrieve the largest or smallest observations, use `nlargest()` and `nsmallest()`.

```python
# Top 5 highest GPAs
df.nlargest(5, "GPA")
```

```python
# Bottom 5 lowest GPAs
df.nsmallest(5, "GPA")
```

- **Counting Values**: The `count()` function counts the number of non-missing values in each column.

```python
df.count()
```

- To count the occurrences of each unique value, use `value_counts()`.

```python
df["Major"].value_counts()
```

```python
# The same method can also be applied to numerical columns.
df["Age"].value_counts()
```

- **Summation**: The `sum()` function computes the total of a numerical column.

```python
print("Total GPA:", df["GPA"].sum())
```

```python
# It can also be applied to multiple numerical columns
# using df.temp()
df_temp[["GPA", "GPA2"]].sum()
```
