### Chapter 8: Data Encoding (Feature Encoding)

Encoding is a technique used to convert categorical data into a numerical format that can be easily understood by machine learning algorithms. This process is essential for preparing data for modeling, as many algorithms require numerical input.

#### Generate Synthetic Data

```python
import numpy as np
import pandas as pd

np.random.seed(42)

names = [
    "James", "Chris", "Adam", "Marian", "Peter",
    "Kevin", "Linda", "Emma", "Sophia", "Daniel",
    "Noah", "Olivia", "Lucas", "Grace", "Henry",
    "Liam", "Ava", "Mia", "Nathan", "Chloe"
]

n = len(names)

df = pd.DataFrame({
    "Name": names,
    "Age": np.random.randint(18, 21, n),
    "Gender": np.random.choice(["Male", "Female"], n),
    "Major": np.random.choice(["AI", "IT", "CSC", "SW"], n),
    "English Level": np.random.choice(["A1", "A2", "B1", "B2"], n),
    "GPA": np.round(np.random.uniform(2.0, 4.0, n), 2)
})

df
```

#### 8.1. Label Encoding

Label Encoding is a method of converting categorical labels into numerical values. Each unique category is assigned a unique integer value. This technique is suitable for ordinal data, where the order of categories matters.

```python
# label encoding can be implemented using map()
df_label_encoded = df.copy()

gender_map = {"Male": 0, "Female": 1}
df_label_encoded["Gender_Label"] = df_label_encoded["Gender"].map(gender_map)

df_label_encoded[["Gender", "Gender_Label"]]
```

```python
major_map = {"AI": 0, "IT": 1, "CSC": 2, "SW": 3}
df_label_encoded["Major_Label"] = df_label_encoded["Major"].map(major_map)

# Remove the Gender and Major column
df_label_encoded = df_label_encoded.drop(columns=['Gender', 'Major'])

df_label_encoded
```

#### 8.2. One-Hot Encoding

One-Hot Encoding is a technique that converts categorical variables into a binary matrix representation. Each category is represented by a binary vector, where only one element is "hot" (1) and the rest are "cold" (0). This method is suitable for nominal data, where the order of categories does not matter.

One-Hot Vector is a representation of categorical variables as binary vectors, where each category is represented by a unique vector with a single "hot" (1) element and the rest as "cold" (0) elements.

```python
major_onehot = pd.get_dummies(df["Major"], 
                              prefix="Major")
major_onehot
```

```python
# This also automatically drops the "Gender" and "Major" columns
df_onehot = pd.get_dummies(df, columns=["Gender", "Major"])
df_onehot
```

```python
# We can replace the True/False with 0/1 using replace(), 
# which is very important for model training
df_onehot = df_onehot.replace({True: 1, False: 0})
df_onehot
```

#### 8.3. Ordinal Encoding

Ordinal Encoding is a technique used to convert categorical variables with a meaningful order into numerical values. Each category is assigned an integer value based on its order or rank. This method is suitable for ordinal data, where the order of categories matters.

```python
df_ordinal_encoded = df.copy()

english_map = {"A1": 1, "A2": 2, "B1": 3, "B2": 4}
df_ordinal_encoded["English_Ordinal"] = df_ordinal_encoded["English Level"].map(english_map)

# Remove the English Level column
df_ordinal_encoded = df_ordinal_encoded.drop(columns=["English Level"])

df_ordinal_encoded
```
