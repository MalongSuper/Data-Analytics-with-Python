## Data Analytics

### Chapter 12: Data Visualization (Statistical Plots)

While categorical plots compare values across groups, **statistical plots** focus on understanding the characteristics of numerical data. They help us examine distributions, identify outliers, explore relationships between variables, and measure correlations.

These visualizations play an important role in **Exploratory Data Analysis (EDA)**, allowing analysts to understand the structure of a dataset before applying statistical methods or machine learning algorithms.

#### Generate Synthetic Data

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(42)
```

```python
n = 200

df = pd.DataFrame({
    "Student ID": [f"S{i:03d}" for i in range(1, n+1)],
    "Age": np.random.randint(18, 24, n),
    "Height": np.random.normal(170, 8, n).round(1),
    "Weight": np.random.normal(65, 12, n).round(1),
    "Gender": np.random.choice(["Male", "Female"], n),
    "Major": np.random.choice(["AI", "IT", "CSC", "SE"], n),
    "GPA": np.round(np.random.uniform(2.0, 4.0, n), 2)
})

df.info()
```

#### 12.1. Scatter Plot

A **scatter plot** visualizes the relationship between two numerical variables by representing each observation as a point. It is commonly used to determine whether two variables exhibit a positive, negative, or no apparent correlation.

```python
sns.scatterplot(
    data=df,
    x="Height",
    y="Weight"
)

plt.title("Height vs Weight")
plt.show()
```

```python
# Height vs Weight by Gender
sns.scatterplot(
    data=df,
    x="Height",
    y="Weight",
    hue="Gender"
)

plt.title("Height vs Weight by Gender")
plt.show()
```

```python
# Scatter Plot with Regression Line
sns.regplot(
    data=df,
    x="Height",
    y="Weight",
    line_kws={'color': 'red'}
)

plt.title("Height vs Weight with Regression Line")
plt.show()
```

#### 12.2. Box Plot

A **box plot** summarizes the distribution of a numerical variable using the five-number summary:

- Minimum (excluding outliers)
- First Quartile (Q1)
- Median
- Third Quartile (Q3)
- Maximum (excluding outliers)

```python
# GPA by Major
sns.boxplot(
    data=df,
    x="Major",
    y="GPA"
)

plt.title("GPA by Major")
plt.show()
```

```python
# GPA by Major and Gender
sns.boxplot(
    data=df,
    x="Major",
    y="GPA",
    hue="Gender"
)

plt.title("GPA by Major and Gender")
plt.show()
```

#### 12.3. Violin Plot

A **violin plot** extends the box plot by displaying the full distribution of the data. In addition to the median and quartiles, it uses **Kernel Density Estimation (KDE)** to show where observations are concentrated.

```python
# GPA by Major
sns.violinplot(
    data=df,
    x="Major",
    y="GPA"
)

plt.title("GPA Distribution by Major")
plt.show()
```

```python
# GPA by Major and Gender
sns.violinplot(
    data=df,
    x="Major",
    y="GPA",
    hue="Gender"
)

plt.title("GPA Distribution by Major and Gender")
plt.show()
```

#### 12.4. Histogram

A **histogram** displays the distribution of a continuous variable by grouping observations into intervals called **bins**.

Histograms help identify characteristics such as:

- Central tendency
- Spread
- Skewness
- Number of peaks (modality)
- Approximate normality

```python
# GPA Distribution
sns.histplot(
    data=df,
    x="GPA"
)

plt.title("Distribution of GPA")
plt.show()
```

```python
# Histogram with KDE
sns.histplot(
    data=df,
    x="GPA",
    kde=True
)

plt.title("Distribution of GPA")
plt.show()
```

#### 12.5. KDE Plot

A **Kernel Density Estimation (KDE) plot** estimates the probability density of a continuous variable using a smooth curve. 

Unlike histograms, KDE plots do not depend on bin sizes, making them especially useful for comparing multiple distributions.

```python
# GPA Distribution
sns.kdeplot(
    data=df,
    x="GPA",
    fill=True
)

plt.title("KDE of GPA")
plt.show()
```

```python
# GPA Distribution by Gender
sns.kdeplot(
    data=df,
    x="GPA",
    hue="Gender",
    fill=True,
    common_norm=False
)

plt.title("GPA Distribution by Gender")
plt.show()
```

#### 12.6. Heatmap

A **heatmap** represents numerical values using color intensity. One of its most common applications is visualizing the **correlation matrix**, where colors indicate the strength and direction of relationships between numerical variables.

```python
# Correlation Heatmap
corr = df.select_dtypes("number").corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Heatmap")
plt.show()
```
