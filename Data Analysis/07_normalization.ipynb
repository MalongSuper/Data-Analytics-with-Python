### Chapter 7: Data Normalization (Feature Scaling)

Normalization is a crucial step in data preprocessing that involves adjusting the values of numeric columns in a dataset to a common scale. 

This process helps improve the performance of machine learning algorithms and ensures that features contribute equally to the model's predictions.

#### Generate Synthetic Data

```python
import numpy as np
import pandas as pd

# Reproducible randomness
np.random.seed(42)

names = [
    "James", "Chris", "Adam", "Marian", "Peter",
    "Kevin", "Linda", "Emma", "Sophia", "Daniel",
    "Noah", "Olivia", "Lucas", "Grace", "Henry",
    "Liam", "Ava", "Mia", "Nathan", "Chloe"
]

majors = ["AI", "IT", "CSC", "SW"]

n = 20

df = pd.DataFrame({
    "Name": names,
    "Gender": np.random.choice(["Male", "Female"], n),
    "Age": np.random.randint(18, 21, n),
    "Height": np.random.randint(150, 191, n),                  # cm
    "Weight": np.random.randint(45, 91, n),                    # kg
    "Major": np.random.choice(majors, n),
    "Scholarship": np.random.choice(["Yes", "No"], n),
    "English Level": np.random.randint(1, 6, n),               # 1-5
    "Study Hours": np.random.randint(5, 31, n),                # hours/week
    "Attendance": np.random.randint(60, 101, n),               # %
    "Distance": np.round(np.random.uniform(0.5, 25, n), 1),    # km
    "Previous Exam": np.random.randint(40, 101, n),            # score
    "GPA": np.round(np.random.uniform(2.0, 4.0, n), 2)
})

df
```

#### 7.1. Retrieving columns based on data types

It is important to identify the numeric columns in a dataset before applying normalization techniques. This can be done by checking the data types of each column and selecting those that are numeric (e.g., integers and floats).

```python
# Select numerical columns
numerical = df.select_dtypes(include=np.number)
print(numerical.columns)
```

Likewise, you can also retrieve non-numeric columns (e.g., categorical columns) for further analysis or preprocessing.

```python
# Select categorical columns
categorical = df.select_dtypes(exclude=np.number)
print(categorical.columns)
```

#### 7.2. Min-Max Normalization

Min-Max normalization is a technique that scales the values of a numeric column to a specified range, typically [0, 1]. 

The formula for Min-Max normalization is:

$$X_{normalized} = \frac{(X - X_{min})}{(X_{max} - X_{min})}$$

Where:
- $X_{normalized}$ is the normalized value,
- $X$ is the original value,
- $X_{min}$ is the minimum value of the column,
- $X_{max}$ is the maximum value of the column.

```python
def min_max_normalize(series):
    return (series - series.min()) / (series.max() - series.min())

# Normalize every numerical column
df_minmax = df.copy()

for col in numerical.columns:
    df_minmax[col] = min_max_normalize(df[col])

df_minmax[numerical.columns]
```

#### 7.3. Z-Score Standardization

Standardization, also known as Z-score normalization, is another technique used to scale numeric columns. It transforms the data to have a mean of 0 and a standard deviation of 1. 

The formula for Z-score standardization is:

$$\Z = \frac{(X - \mu)}{\sigma}$$

Where:
- $Z$ is the standardized value,
- $X$ is the original value,
- $\mu$ is the mean of the column,
- $\sigma$ is the standard deviation of the column.

```python
def z_score_normalize(series):
    return (series - series.mean()) / series.std()

df_zscore = df.copy()

for col in numerical.columns:
    df_zscore[col] = z_score_normalize(df[col])

df_zscore[numerical.columns]
```

#### 7.4. Log Transformation

Log transformation is a technique used to reduce the skewness of a distribution and make it more symmetric. It is particularly useful for data that spans several orders of magnitude. The log transformation can be applied using the natural logarithm (ln) or base 10 logarithm (log10). 

The formula for log transformation is:

$$X_{log} = \log(X + 1)$$

Where:
- $X_{log}$ is the log-transformed value,
- $X$ is the original value (adding 1 to avoid taking the log of zero).

```python
def log_transform(series):
    return np.log1p(series)

df_log = df.copy()

# Only apply to positive-valued columns
for col in numerical.columns:
    if (df[col] >= 0).all():
        df_log[col] = log_transform(df[col])

df_log[numerical.columns]
```

#### 7.5. Robust Scaling

Robust scaling is a technique that uses the median and the interquartile range (IQR) to scale numeric columns. This method is less sensitive to outliers compared to Min-Max normalization and Z-score standardization. The formula for robust scaling is:

$$X_{robust} = \frac{(X - \text{median})}{\text{IQR}}$$

Where:
- $X_{robust}$ is the robust-scaled value,
- $X$ is the original value,
- $\text{median}$ is the median of the column,
- $\text{IQR}$ is the interquartile range of the column (Q3 - Q1).

```python
def robust_scale(series):
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    return (series - series.median()) / iqr

df_robust = df.copy()

for col in numerical.columns:
    df_robust[col] = robust_scale(df[col])

df_robust[numerical.columns]
```

#### 7.6. Max Absolute Scaling

Max Absolute Scaling is a technique that scales the values of a numeric column by dividing each value by the maximum absolute value of that column. This method is particularly useful when the data contains both positive and negative values. 

The formula for Max Absolute Scaling is:

$$X_{max\_abs} = \frac{X}{|X_{max}|}$$

Where:
- $X_{max\_abs}$ is the max absolute scaled value,
- $X$ is the original value,
- $|X_{max}|$ is the maximum absolute value of the column.

```python
def max_absolute_scale(series):
    return series / series.abs().max()

df_maxabs = df.copy()

for col in numerical.columns:
    df_maxabs[col] = max_absolute_scale(df[col])

df_maxabs[numerical.columns]
```
