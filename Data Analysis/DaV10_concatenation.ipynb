## Data Analytics

### Chapter 10: Data Combination (Concatenation, Comparison, and Segmentation)

- Concatenating DataFrames combines multiple DataFrames by stacking them vertically (adding rows) or horizontally (adding columns). Pandas provides the `pd.concat()` function to combine multiple DataFrames.

- Comparing DataFrames identifies the differences between two similar DataFrames, making it useful for validating updates or detecting changes after data preprocessing.

- Segmentation groups continuous numerical values into categorical intervals using pd.cut(), allowing numerical features to be represented as meaningful categories for analysis or machine learning.

#### Generate Synthetic Data

```python
import numpy as np
import pandas as pd

np.random.seed(42)
```

```python
studentAI = pd.DataFrame({
    "Name": [f"AI_{i}" for i in range(1,21)],
    "Age": np.random.randint(18,22,20),
    "GPA": np.round(np.random.uniform(2.0,4.0,20),2)
})


studentAI
```

```python
studentCSC = pd.DataFrame({
    "Name": [f"CSC_{i}" for i in range(1,21)],
    "Age": np.random.randint(18,22,20),
    "GPA": np.round(np.random.uniform(2.0,4.0,20),2)
})

studentCSC
```

#### 10.1. Vertical Concatenation

Row-wise (Vertical): Stack one dataset below another. This is commonly used when combining datasets with identical columns.

```python
result = pd.concat(
    [studentAI, studentCSC],
    axis=0,
    ignore_index=True
)

result
```

#### 10.2. Horizontal Concatenation

Column-wise (Horizontal): Place datasets side-by-side. Unlike merge(), the DataFrames do not need a common key, but they should have the same number of rows or compatible indexes.

```python
result = pd.concat(
    [studentAI, studentCSC],
    axis=1
)

result
```

#### 10.3. Comparing Two DataFrames

During data cleaning or preprocessing, it is common to compare two similar DataFrames to determine which values have changed. Pandas provides the `compare()` method to highlight these differences.

```python
studentAI2 = studentAI.copy()

studentAI2.loc[0, "GPA"] = 3.95
studentAI2.loc[3, "Age"] = 21
studentAI2.loc[5, "GPA"] = 2.50

comparison = studentAI.compare(studentAI2)

comparison
```

#### 10.4. Data Segmentation

Sometimes it is useful to convert continuous numerical values into categorical groups. This process is known as **segmentation** or **binning**.

Pandas provides the **`pd.cut()`** function to divide numerical values into intervals.

```python
# Segmenting Age
studentAI["Age Group"] = pd.cut(
    studentAI["Age"],
    bins=[17,18,20,22],
    labels=["18","19-20","21"]
)

studentAI
```

```python
# Segmenting GPA
studentAI["Performance"] = pd.cut(
    studentAI["GPA"],
    bins=[0,2.5,3.2,4.0],
    labels=["Low","Average","High"]
)

studentAI
```
