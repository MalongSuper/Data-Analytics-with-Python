# Data Visualization Guide: Choosing the Right Chart

We explore various types of data plots and visual representations.

### Why Choosing the Right Visualization Matters
Selecting the correct visualization is critical because different charts serve different analytical purposes:
- **Line Charts**: Excellent for displaying trends and changes over continuous time intervals (e.g., monthly revenue).
- **Scatter Plots**: Ideal for showing the relationship or correlation between two numerical variables.
- **Histograms**: Great for understanding the distribution and frequency of a single continuous variable.
- **Pie Charts**: Useful for comparing parts of a whole, though best kept to a small number of categories.
- **Dual-Axis Charts**: Helpful when comparing two different metrics (like revenue and profit margins) that share a common dimension but use different scales.

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
revenue = [12000, 13500, 12800, 16000, 17500, 21000]

plt.plot(months, revenue, marker='o')
plt.title('Monthly Revenue')
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Seed for reproducibility
np.random.seed(42)

# Create 20 samples of mock data
data = {
    'Advertising_Cost': np.random.randint(100, 1000, 20),
    'Revenue': np.random.randint(5000, 25000, 20),
    'Department': np.random.choice(['IT', 'Business', 'Science', 'Marketing'], 20)
}
df = pd.DataFrame(data)

sns.scatterplot(
    data=df,
    x='Advertising_Cost',
    y='Revenue',
    hue='Department'
)

plt.title('Advertising Cost vs Revenue')
plt.show()
```

```python
# Generate a simulated Customer_Age column for the 20 samples
np.random.seed(42)
df['Customer_Age'] = np.random.randint(18, 70, len(df))

sns.histplot(data=df, x='Customer_Age', bins=15, kde=True)
plt.title('Distribution of Customer Age')
plt.show()
```

```python
categories = ['Electronics', 'Clothing', 'Food', 'Furniture']
revenue = [45000, 25000, 18000, 12000]

plt.pie(
    revenue,
    labels=categories,
    autopct='%1.1f%%',
    startangle=90
)
plt.title('Revenue Share by Category')
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
revenue = [120, 150, 140, 180, 210]
profit_margin = [12, 15, 13, 18, 20]

fig, ax1 = plt.subplots(figsize=(8, 4))

ax1.bar(months, revenue, label='Revenue')
ax1.set_ylabel('Revenue')

ax2 = ax1.twinx()
ax2.plot(
    months, profit_margin,
    color='red', marker='o',
    label='Profit Margin'
)
ax2.set_ylabel('Profit Margin (%)')

plt.title('Revenue and Profit Margin')
plt.show()
```
