# Pandas

Pandas is a fast, powerful, and easy-to-use open-source data analysis and manipulation library built on top of the Python programming language.

*Overview and Core Features*

* **Definition:** Pandas gets its name from "panel data", an econometrics term for multi-dimensional data sets.

* **Data Structures:** It relies on two main structures: Series (a one-dimensional labeled array) and DataFrame (a two-dimensional labeled table with rows and columns).

* **Data Operations:** It allows you to clean, filter, group, merge, and reshape data smoothly.

* **File Support**: You can easily import and export data from formats like CSV, Excel, JSON, and SQL databases.

### Pandas Series

A Pandas Series is a one-dimensional labeled array capable of holding data of any type (integers, strings, floating-point numbers, Python objects, etc.). You can think of it as a single column in an Excel spreadsheet or a database table.

A collection of these Series aligned by a common index forms a DataFrame (a two-dimensional table)

```python
# Series from list
data = [1, 2, 3, 4, 5]

series1 = pd.Series(data)
series1
```

```python
# Series from list with different data types
new_data = ['James', 'A', 12, True]

series2 = pd.Series(new_data)
series2
```

```python
# Creating a series of name
data = ['James', 'Chris', 'Adam', 'Jessie', 'Alice']

series = pd.Series(data)
series
```

```python
# Series with value name
data = ['James', 'Chris', 'Adam', 'Jessie', 'Alice']

series = pd.Series(data, name='Employee')
series
```

```python
# Series with index name
data = ['James', 'Chris', 'Adam', 'Jessie', 'Alice']

series = pd.Series(data, index=['a', 'b', 'c', 'd', 'e'],
                   name='Employee')
series
```

### Pandas DataFrames

A Pandas DataFrame is a two-dimensional, size-mutable, and potentially heterogeneous tabular data structure with labeled axes (rows and columns). Think of it like a SQL table, a Microsoft Excel spreadsheet, or a dictionary containing Pandas Series objects. It is the core data structure used in the Python Pandas Library for data manipulation and analysis

```python
# Returns GPA greater than or equal to 3.0
df[df['GPA'] >= 3.0]
```

```python
# Returns those whose age is smaller than the mean
df[df['Age'] < df['Age'].mean()]
```

```python
# Major is AI
df[df['Major'] == 'AI']
```

```python
# Returns IT Student
# whose GPA greater than or equal to 3.0
df[(df['Major'] == 'IT') & (df['GPA'] >= 3.0)]
```

```python
# Returns AI Student or Student with Age above 25
df[(df['Major'] == 'AI') | (df['Age'] > 25)]
```

```python
# Return non-IT Student
df[~(df['Major'] == 'IT')]
```

```python
# Return a semester record that exists
# a GPA score that is above 3.5 -> any()
gpa_cols = ['GPA', 'GPA_2', 'GPA_3']
df[(df[gpa_cols] > 3.5).any(axis=1)]
```

```python
# Return a semester record that exists
# every GPA score that is above 2.4 -> all()
gpa_cols = ['GPA', 'GPA_2', 'GPA_3']
df[(df[gpa_cols] > 2.4).all(axis=1)]
```

```python
students_dict = {
    "Student": ["Alice", "Bob", "Charlie", "Diana", "Ethan"],
    "Course": ["CS101", "MATH201", "BIO150", "CS101", "ENG210"],
    "GPA": [3.5, 3.8, 2.9, 3.2, 3.7]
}

courses_dict = {
    "Course": ["CS101", "MATH201", "BIO150", "ENG210", "HIST110"],
    "Major": ["Computer Science", "Mathematics", "Biology",
              "English", "History"]
}

df1 = pd.DataFrame(students_dict)
df2 = pd.DataFrame(courses_dict)

# Merge dataframes
df_merged = pd.merge(df1, df2, on='Course')
print(df_merged)
```

```python
# Join dataframes
df1 = df1.set_index('Course')
df2 = df2.set_index('Course')
df_joined = df1.join(df2)
print("\n", df_joined)
```

```python
df_merged = pd.merge(df1, df2, on='Course')

# Group by Major, aggregate on GPA mean
df_grouped = df_merged.groupby('Major')['GPA'].mean()
print(df_grouped)
```

```python
# Concatenation
df1 = pd.DataFrame(students_dict)
df3 = pd.DataFrame({'Age': [18, 19, 19, 19, 18]})
pd.concat([df1, df3], axis=1)
```

```python
# Segmentation based on GPA (>= 3.0 and < 3.0)
# pd.cut()
bins = [0, 3.0, 4.0]  # GPA ranges
labels = ["< 3.0", ">= 3.0"]
df1["GPA"] = pd.cut(df1["GPA"], bins=bins,
                    labels=labels, right=False)
df1
```
