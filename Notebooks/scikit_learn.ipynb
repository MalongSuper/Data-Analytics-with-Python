Scikit-learn (also known as sklearn) is the most popular open-source machine learning library for Python. It provides a robust, user-friendly toolset for building predictive models and performing statistical modeling.

Built on top of Python's fundamental scientific stack—NumPy, SciPy, and Matplotlib—it integrates seamlessly with data manipulation libraries like Pandas.

*Core Pillars of Scikit-Learn*

The library organizes its massive suite of tools into six primary categories:

* **Classification:** Identifying which category an object belongs to (e.g., spam filtering, image recognition, or customer churn).

* **Regression:** Predicting a continuous-valued numeric attribute (e.g., housing prices or stock trends).

* **Clustering:** Automatically grouping similar data points together without labels (e.g., customer segmentation).

* **Dimensionality Reduction:** Reducing the number of random variables to analyze (e.g., PCA for data compression or visualization).

* **Preprocessing:** Transforming raw feature vectors into representations suited for machine learning algorithms (e.g., text encoding or feature scaling).

* **Model Selection:** Comparing, validating, and choosing optimal parameters using utilities like cross-validation and grid search

```python
# Read the data
data = {
    "Student Name": [
        "Chris C", "John A", "Alice B", "Emma D", "David E",
        "Sophia F", "Michael G", "Olivia H", "Daniel I", "Mia J"
    ],
    "Major": [
        "Computer Science",
        "Business",
        "Engineering",
        "Engineering",
        "Engineering",
        "Business",
        "Computer Science",
        "Business",
        "Business",
        "Computer Science"
    ],
    "Academic Level": [
        "High", "Medium", "Low", "High", "Medium",
        "Low", "High", "Medium", "Low", "High"
    ],
    "Occupation": [
        "Student", "Worker", "Student", "Student", "Worker",
        "Other", "Student", "Worker", "Other", "Student"
    ]
}

df = pd.DataFrame(data)
df
```

```python
# One-Hot Encoding the Major column
encoder = OneHotEncoder(sparse_output=False)
# This returns only a matrix
encoded_data = encoder.fit_transform(df[['Major']])
# Get the feature names
encoded_cols = encoder.get_feature_names_out(['Major'])
print(encoded_cols)
# We need to read the matrix as DataFrame
encoded_df = pd.DataFrame(encoded_data, columns=encoded_cols)
print("\n", encoded_df)

# Concatenate (drop te majors)
df = pd.concat([df.drop('Major', axis=1), encoded_df], axis=1)
df
```

```python
print("Unique Labels:", df['Occupation'].unique())
# Label Encoding the Occupation column
encoder = LabelEncoder()
df['Occupation'] = encoder.fit_transform(df['Occupation'])
df
```

```python
print("Unique Labels:", df['Academic Level'].unique())
# Ordinal Encoding the Academic Level column
# The order of the kist determines the ranking of the categories
encoder = OrdinalEncoder(categories=[['Low', 'Medium', 'High']])
# This also just returns a matrix
encoded_data = encoder.fit_transform(df[['Academic Level']])
encoded_cols = encoder.get_feature_names_out(['Academic Level'])
print(encoded_cols)

# We need to read the matrix as DataFrame
encoded_df = pd.DataFrame(encoded_data, columns=encoded_cols)
print("\n", encoded_df)

# Concatenate
df = pd.concat([df.drop('Academic Level', axis=1), encoded_df], axis=1)
df
```

## **Imputers**

First, let's load the diabetes dataset and intentionally introduce some missing values to demonstrate imputers.

```python
# Load the diabetes dataset
diabetes = datasets.load_diabetes()

# Read the dataset into a DataFrame
df_diabetes = pd.DataFrame(diabetes.data, columns=diabetes.feature_names)
df_diabetes['target'] = diabetes.target

# Introduce 20% missing values randomly
# Calculate the total number of cells in the DataFrame
total_cells = df_diabetes.size

# Calculate the number of cells to turn into NaN (20%)
num_missing = int(total_cells * 0.2)

# Get random row and column indices
missing_row_indices = np.random.randint(0, df_diabetes.shape[0], num_missing)
missing_col_indices = np.random.randint(0, df_diabetes.shape[1], num_missing)

# Set values to NaN
for i in range(num_missing):
    df_diabetes.iloc[missing_row_indices[i], missing_col_indices[i]] = np.nan

print("DataFrame with 20% missing values:")
df_diabetes.head()
```

```python
# Imputed with median
# 'mean' is the default behavior, others are median and mode
imputer = SimpleImputer(strategy='median')
# Create a new dataframe
df_imputed1 = pd.DataFrame(imputer.fit_transform(df_diabetes), columns=df_diabetes.columns)
df_imputed1.head()
```

```python
# KNN Imputer: Using nearest neighbors
imputer = KNNImputer(n_neighbors=5)
df_imputed2 = pd.DataFrame(imputer.fit_transform(df_diabetes), columns=df_diabetes.columns)
df_imputed2.head()
```

```python
# MICE Method
imputer = IterativeImputer(max_iter=10, random_state=0)
df_imputed3 = pd.DataFrame(imputer.fit_transform(df_diabetes), columns=df_diabetes.columns)
df_imputed3.head()
```

## **Model Training & Evaluation**

**Supervised and Unsupervised Models Overview**

**Supervised**
| **Model Type (Module)** | **Model Name** | **What the Model Does** | **Implementation** |
| --- | --- | --- | --- |
| Linear Models (sklearn.linear_model) | Linear Regression | The simplest regression model, finds a linear relationship between features and target | `LinearRegression()` |
| Linear Models (sklearn.linear_model) | Ridge / Lasso Regression | Linear models with L2 (Ridge) or L1 (Lasso) regularization to prevent overfitting | `Ridge()` / `Lasso()` |
| Linear Models (sklearn.linear_model) | Logistic Regression | Classification model using a sigmoid curve to separate classes | `LogisticRegression()` |
| Tree-Based Models (sklearn.tree) | Decision Tree | Splits data into decisions for classification or regression | `DecisionTreeClassifier()` / `DecisionTreeRegressor()` |
| Ensemble Models (sklearn.ensemble) | Random Forest | Combines multiple decision trees to improve performance | `RandomForestClassifier()` / `RandomForestRegressor()` |
| Ensemble Models (sklearn.ensemble) | Gradient Boosting | Builds trees sequentially, correcting previous errors | `GradientBoostingClassifier()` / `GradientBoostingRegressor()` |
| Probabilistic Models (sklearn.naive_bayes) | Naïve Bayes | Uses probability assumptions for fast classification | `GaussianNB()` / `MultinomialNB()` / `CategoricalNB()` |
| Discriminant Analysis (sklearn.discriminant_analysis) | Linear Discriminant Analysis (LDA) | Reduces dimensions while separating classes | `LinearDiscriminantAnalysis()` |
| Discriminant Analysis (sklearn.discriminant_analysis) | Quadratic Discriminant Analysis (QDA) | Similar to LDA but allows more flexible boundaries | `QuadraticDiscriminantAnalysis()` |
| Nearest Neighbors (sklearn.neighbors) | K-Nearest Neighbors | Predicts based on similarity to nearest data points | `KNeighborsClassifier()` / `KNeighborsRegressor()` |
| Support Vector Machines (sklearn.svm) | Support Vector Machine | Finds optimal hyperplane to separate data | `SVC()` / `SVR()` |
| Neural Networks (sklearn.neural_network) | Multi-Layer Perceptron | A basic neural network for classification or regression | `MLPClassifier()` / `MLPRegressor()` |

**Unsupervised**
| **Model Type (Module)** | **Model Name** | **What the Model Does** | **Implementation** |
| --- | --- | --- | --- |
| Clustering (sklearn.cluster) | K-Means | Divides data into K clusters based on similarity | `KMeans()` |
| Clustering (sklearn.cluster) | DBSCAN | Groups data based on density and detects noise | `DBSCAN()` |
| Dimensionality Reduction (sklearn.decomposition) | Principal Component Analysis (PCA) | Reduces features while preserving variance | `PCA()` |
| Dimensionality Reduction (sklearn.manifold) | t-SNE | Projects high-dimensional data into lower dimensions for visualization | `TSNE()` |

```python
# Import train_test_split
from sklearn.model_selection import train_test_split

# Import machine learning models
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LinearRegression
from sklearn.cluster import KMeans
from sklearn.ensemble import RandomForestClassifier

# Import metrics
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                             f1_score, confusion_matrix, classification_report)
from sklearn.metrics import mean_squared_error, r2_score

# Import visualization tool
import seaborn as sns
import matplotlib.pyplot as plt
```

Model Training with K-Nearest Neighbors (Iris Dataset)

```python
# The iris dataset
iris = datasets.load_iris()

# Read it as DataFrame
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['Target'] = iris.target

#Target and features
target = 'Target'
features = df.columns.drop(target)

#Split the data into 80% for training and 20% for testing
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42)

X_train = train_df[features]
y_train = train_df[target]
X_test = test_df[features]
y_test = test_df[target]

# Model Training with KNN
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)

# Model Evaluation
y_pred = knn.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
class_report = classification_report(y_test, y_pred)

print(f"Accuracy: {accuracy}")
print(f"Classification Report:\n{class_report}")
```

Model Training with Decision Tree (Breast Cancer Dataset)

```python
# The breast cancer dataset
bc_data = datasets.load_breast_cancer()

# Read it as DataFrame
df = pd.DataFrame(data=bc_data.data, columns=bc_data.feature_names)
df['Target'] = bc_data.target

#Target and features
target = 'Target'
features = df.columns.drop(target)

# Split the data into 80% for training and 20% for testing
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42)
X_train = train_df[features]
y_train = train_df[target]
X_test = test_df[features]
y_test = test_df[target]

# Model Training with Decision Tree
model = DecisionTreeClassifier(random_state=42)
model.fit(X_train, y_train)

# Model Evaluation
# Default is for binary classification
# For multi-class, you must use average=macro
y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1_score = f1_score(y_test, y_pred)

# Print the results
print(f"Accuracy: {accuracy}")
print(f"Precision: {precision}")
print(f"Recall: {recall}")
print(f"F1 Score: {f1_score}")
```

```python
# Plot confusion matrix
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
```

```python
# Plot the metrics as bar plot
sns.barplot(x=['Accuracy', 'Precision', 'Recall', 'F1 Score'],
        y=[accuracy, precision, recall, f1_score],
        palette=['red', 'blue', 'green', 'purple'])

for i, v in enumerate([accuracy, precision, recall, f1_score]):
    plt.text(i, v + 0.02, f"{v:.4f}", ha='center')
```

Diabetes dataset with Linear Regression

```python
# Load the diabetes dataset
diabetes = datasets.load_diabetes()

# Read it as DataFrame
df = pd.DataFrame(data=diabetes.data, columns=diabetes.feature_names)
df['Target'] = diabetes.target

# Target and features
target = 'Target'
features = df.columns.drop(target)

# Split the data into 80% for training and 20% for testing
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42)

X_train = train_df[features]
y_train = train_df[target]
X_test = test_df[features]
y_test = test_df[target]

# Model Training with Linear Regression
model = LinearRegression()
model.fit(X_train, y_train)

# Model Evaluation
y_pred = model.predict(X_test)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"Mean Squared Error: {mse}")
print(f"R-squared: {r2}")

# Multiple Regression
print("Intercept:", model.intercept_)
print("Coefficients:", model.coef_)
```

K-Means Clustering

```python
# Load the iris dataset for clustering
iris = datasets.load_iris()

# Read it as DataFrame (No target variable needed for unsupervised learning)
X = pd.DataFrame(data=iris.data, columns=iris.feature_names)

# Model Training with K-Means (we choose 3 clusters)n
kmeans = KMeans(n_clusters=3)
kmeans.fit(X)

# Display for Sepal
plt.scatter(X['sepal length (cm)'], X['sepal width (cm)'], c=kmeans.labels_)
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.title('K-Means Clustering (Sepal Features)')
plt.show()

# Display for Petal
plt.scatter(X['petal length (cm)'], X['petal width (cm)'], c=kmeans.labels_)
plt.xlabel('Petal Length')
plt.ylabel('Petal Width')
plt.title('K-Means Clustering (Petal Features)')
plt.show()
```

Random Forest with Feature Importance

```python
# The breast cancer dataset
wine = datasets.load_wine()

# Read it as DataFrame
df = pd.DataFrame(data=wine.data, columns=wine.feature_names)
df['Target'] = wine.target

# Target and features
target = 'Target'
features = df.columns.drop(target)

# Split the data into 80% for training and 20% for testing
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42)
X_train = train_df[features]
y_train = train_df[target]
X_test = test_df[features]
y_test = test_df[target]

# Train with Random Forest (100 estimators/trees)
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# Plot feature importance
importances = model.feature_importances_
indices = np.argsort(importances)[::-1]

plt.figure(figsize=(10, 6))
plt.title("Feature Importance")
plt.barh(range(X_train.shape[1]), importances[indices], align='center')
plt.yticks(range(X_train.shape[1]), np.array(features)[indices])
plt.xlabel("Relative Importance")
plt.show()
```
