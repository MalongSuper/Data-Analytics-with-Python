Seaborn is a popular Python data visualization library built on top of Matplotlib that makes it easy to draw attractive and informative statistical graphics.

*Key Features*

* **Pandas Integration:** Works directly with Pandas DataFrames by accepting column names as arguments.

* **Statistical Aggregation: **Automatically computes mappings, groupings, and statistical estimations (like bootstrapping confidence intervals) behind the scenes.

* **Built-in Themes:** Offers clean default styles and attractive color palettes that elevate standard Matplotlib visuals instantly.

*Core Plot Types*

* **Relational Plots:** Show relationships between two continuous variables using scatterplot or lineplot (e.g., time series).

* **Categorical Plots:** Deal with discrete groups using boxplots, violin plots, or bar plots.

* **Distribution Plots:** Visualize univariate or bivariate distributions using histograms (histplot) or kernel density estimates (kdeplot).

* **Matrix Plots:** Represent matrix-like or correlation data using heatmaps

### Install and Import the library

```python
# !pip install pandas seaborn
import seaborn as sns
import pandas as pd
import os
```

#### Load the Dataset

Original dataset link:
https://www.kaggle.com/datasets/mahmoudelhemaly/students-grading-dataset?resource=download

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("mahmoudelhemaly/students-grading-dataset")

print("Path to dataset files:", path)
```

```python
files = os.listdir(path)
print("Files in directory:", files)

# Read the CSV file into a DataFrame
# (Assuming the file is named 'students_grading_dataset.csv' or similar)
csv_file = [f for f in files if f.endswith('.csv')][0]
df = pd.read_csv(os.path.join(path, csv_file))

# View the first few rows
df.head(10)
```

```python
# Retrieve some information
df.info()
```

```python
# For demonstration, we only use the first 100 samples
df = df.head(100)
```

#### Data Visualization with Seaborn

```python
# Scatter Plot
# Relationship between two numerical variables
sns.scatterplot(data=df, x="Study_Hours_per_Week", y="Total_Score")
```

```python
# Line Plot
# Relationship across an ordered variable
# hue= separates data into groups with the given column
sns.lineplot(data=df, x="Age", y="Total_Score",
             hue="Department")
```

```python
# Bar plot
# Show the value of y based on category x
sns.barplot(data=df, x="Department", y="Total_Score")
```

```python
# Box Plot: Show median and quartiles
# The line represents the median (second quartile)
# The box represents first and third quartiles
sns.boxplot(data=df, x="Gender", y="Final_Score")
```

```python
# Violin Plot: Similar to boxplot
# but shows the whole probability distribution
sns.violinplot(data=df, x="Family_Income_Level", y="Total_Score")
```

```python
# Strip Plot: Shows each data point on the graph
# jitter=True: for overlapping pojts, add a small offset to spread them out
sns.stripplot(data=df, x="Gender", y="Study_Hours_per_Week", jitter=True)
```

```python
# Swarm Plot
# Improving Strip Plot without overlapping points
sns.swarmplot(data=df, x="Department", y="Attendance (%)")
```

```python
# Point Plot
# Shows summary statistic (mean) with confidence interval
sns.pointplot(data=df, x="Gender", y="Total_Score", hue="Department")
```

```python
# Regression Plot
# Shows the relationship between two numeric variables
# Fits a linear regression line and confidence interval
sns.regplot(data=df, x="Study_Hours_per_Week", y="Total_Score")
```

```python
# Linear Model Plot
# Similar to Regression Plot but with row, col, hue
sns.lmplot(data=df, x="Study_Hours_per_Week", y="Total_Score",
           hue="Gender", col="Department", height=4)
```

```python
# Heatmap: Correlation
numeric_df = df[['Midterm_Score', 'Final_Score',
                 'Assignments_Avg', 'Quizzes_Avg',
                 'Participation_Score', 'Projects_Score', 'Total_Score']]

sns.heatmap(numeric_df.corr(), annot=True, fmt=".2f", cmap="coolwarm")
```

```python
# Catplot: Combine categorical statistics with faceting
sns.catplot(data=df, x="Family_Income_Level",
            y="Total_Score", hue="Gender",
            kind="point", col="Department")
```

```python
# Histogram: Shows the frequency of values
sns.histplot(data=df, x="Total_Score", bins=20)
```

```python
# Kernel Density Estimate
# Smooth curve representing the distribution
sns.kdeplot(data=df, x="Total_Score", fill=True)
```

```python
# Rug Plot: Shows each individual data point along an axis
# Often combined with Histogram and KDE
# Histogram with KDE
sns.histplot(data=df, x="Total_Score", bins=10, kde=True)
# Add rug plot on the same axis
sns.rugplot(data=df, x="Total_Score", height=0.05, color='red')
```

```python
# Joint Plot is a hybrid plot
# Shows Relational and distribution in one figure
sns.jointplot(x='Study_Hours_per_Week', y='Total_Score', data=df)
```

```python
# Same joint plot but with KDE
sns.jointplot(x='Study_Hours_per_Week', y='Total_Score', data=df, kind='kde')
```
