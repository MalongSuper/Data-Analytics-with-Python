## SciPy

SciPy is an open-source Python library used for scientific, engineering, and technical computing. It is built directly on top of NumPy and extends it with tools for calculus, signal processing, optimization, statistics, and more.

While NumPy handles efficient array manipulation, SciPy provides the heavy-lifting tools used for scientific computing.

Because its core algorithms are implemented in C, C++, and Fortran, SciPy can execute complex calculations much faster than pure Python code.

```python
import numpy as np
from scipy.sparse import coo_matrix, csr_matrix, csc_matrix

data = np.array([[0, 1, 0], [0, 0, 2], [3, 0, 4]])

sparse_coo = coo_matrix(data)
print('COO matrix:')
print(sparse_coo)
print('Row indices:', sparse_coo.row)
print('Column indices:', sparse_coo.col)
print('Values:', sparse_coo.data)

sparse_csr = csr_matrix(data)
print('CSR matrix:')
print(sparse_csr)
print('Row pointers:', sparse_csr.indptr)
print('Column indices:', sparse_csr.indices)
print('Values:', sparse_csr.data)

data2 = np.array([[0, 1, 0], [0, 5, 2], [3, 0, 4]])
sparse_csc = csc_matrix(data2)
print('CSC matrix:')
print(sparse_csc)
print('Column pointers:', sparse_csc.indptr)
print('Row indices:', sparse_csc.indices)
print('Values:', sparse_csc.data)
```

```python
import numpy as np
from scipy.sparse import csr_matrix

data1 = np.array([[0, 1, 5], [0, 0, 2], [0, 0, 4]])
sparse_csr1 = csr_matrix(data1)

print('Number of non-zeros:', sparse_csr1.count_nonzero())
print('Non-zero counts by columns:', sparse_csr1.getnnz(axis=0))
print('Non-zero counts by rows:', sparse_csr1.getnnz(axis=1))
print('Shape:', sparse_csr1.shape)
print('Data shape:', sparse_csr1.data.shape)

data2 = np.array([[1, 1, 5], [1, 4, 5], [2, 0, 2]])
sparse_csr2 = csr_matrix(data2)
print('Addition:')
print(sparse_csr1 + sparse_csr2)
print('Subtraction:')
print(sparse_csr1 - sparse_csr2)
print('Dot product:')
print(sparse_csr1.dot(sparse_csr2))
```

```python
import numpy as np
from scipy.sparse import csr_matrix

data1 = np.array([[0, 1, 5], [0, 0, 2], [0, 0, 4]])
sparse_csr1 = csr_matrix(data1)

print('Mean of each column:', sparse_csr1.mean(axis=0))
print('Sum of each column:', sparse_csr1.sum(axis=0))
print('Min of each column:', sparse_csr1.min(axis=0))
print('Max of each column:', sparse_csr1.max(axis=0))
print('Mean of each row:', sparse_csr1.mean(axis=1))
print('Sum of each row:', sparse_csr1.sum(axis=1))
print('Min of each row:', sparse_csr1.min(axis=1))
print('Max of each row:', sparse_csr1.max(axis=1))
```

```python
import scipy.stats as stats
import numpy as np

data = np.array([1, 2, 3, 2, 4, 6, 3, 2, 8, 8])
print('Describe:')
print(stats.describe(data))
print('Mode:')
print(stats.mode(data))

data2 = np.array([[1, 2, 2], [2, 4, 4], [3, 3, 9]])
print('Mode by axis=0:', stats.mode(data2, axis=0))
print('Mode by axis=1:', stats.mode(data2, axis=1))
```

```python
import scipy.stats as stats
import numpy as np

data = np.array([1, 2, 3, 2, 4, 6, 3, 2, 8, 12, 99, 13, 11])
print('Mean:', np.mean(data), 'Trimmed Mean:', stats.tmean(data, limits=(1, 20)))
print('Variance:', np.var(data), 'Trimmed Variance:', stats.tvar(data, limits=(1, 20)))
print('Std:', np.std(data), 'Trimmed Std:', stats.tstd(data, limits=(1, 20)))
print('Min, Max:', np.min(data), np.max(data), 'Trimmed Min, Max:', stats.tmin(data), stats.tmax(data))
print('Skewness:', stats.skew(data))
print('Kurtosis:', stats.kurtosis(data))
```

```python
from scipy import stats

ber = stats.bernoulli(p=0.6)
print('P(X = 1):', ber.pmf(1))
print('P(X <= 2):', ber.cdf(2))
print('Stats:', ber.mean(), ber.var(), ber.std())

bino = stats.binom(10, 0.6)
print('P(X = 5):', bino.pmf(5))
print('P(X >= 3):', bino.sf(2))
print('P(X <= 4):', bino.cdf(4))
print('Mean:', bino.mean(), 'Variance:', bino.var(), 'Std:', bino.std())

mu = 4
pois = stats.poisson(mu)
print('P(X = 5):', pois.pmf(5))
print('P(X > 5):', pois.sf(5))
print('P(X < 4):', pois.cdf(3))

normal = stats.norm(6, 2)
print('P(X < 10):', normal.cdf(10))
print('P(X > 4):', normal.sf(4))
print('PDF at X=4:', normal.pdf(4))
print('Mean:', normal.mean(), 'Variance:', normal.var(), 'Std:', normal.std())
```

```python
from scipy import stats
import numpy as np

data = [2.5, 2.0, 3.5, 5.0, 4.0, 7.2, 4.5, 9.8, 2.5, 3.5, 2.2, 4.5, 2.5, 3.0, 1.5, 3.0, 1.5, 3.5, 3.0, 2.1, 4.5]
z_scores = stats.zscore(data)
print('Z-scores:', z_scores)

X = np.array([80, 82, 85, 90, 100, 110, 120, 150])
Y = np.array([2.5, 2.7, 2.8, 2.9, 3.0, 3.1, 3.2, 3.3])
linregress = stats.linregress(X, Y)
print(f'Y = {linregress.intercept:.4f} + {linregress.slope:.4f}X')
```
